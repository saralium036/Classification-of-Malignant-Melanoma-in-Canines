### Dealing with class imbalance

As seen from the plot above, there is an imbalance in the class distribution, I will use compute class weight and SMOTE to deal with the imbalance
#### Compute Class Weight

```python
from sklearn.utils.class_weight import compute_class_weight
# classes from the y labels
class_values = np.unique(np.argmax(y_train, axis = 1))
class_weights = compute_class_weight('balanced',
                     classes = class_values,
                    y = np.argmax(y_train, axis = 1) )
class_weight = dict(zip(class_values, class_weights))
```

#### Sample Minority Oversampling Technique (SMOTE)

```python
#Import the libraries needed
from sklearn.utils import shuffle
from imblearn.over_sampling import SMOTE

# First since the images are 3d, i have to Flatten each image to a 1D array
X_train_flat = X_train.reshape(X_train.shape[0], -1)

# Ensure y_train is a 1D array
y_train_flat = np.argmax(y_train, axis = 1)

# Use SMOTE on the flattened data
smote = SMOTE(random_state=42)
X_train_resampled, y_train_resampled = smote.fit_resample(X_train_flat, y_train_flat)

# Reshape the data back to 4D if needed
X_train_resampled = X_train_resampled.reshape(-1, 224, 224, 3)

# Now X_train_resampled and y_train_resampled contain the resampled data
```

```python
y_train_resampled = np_utils.to_categorical(y_train_resampled, 5)
```

### Building the Model

```python

# I will use define function to build my model and call it LCNet. ideas for the architecture is from Kaur et al., (2021)
def lcnet():
    model = Sequential()

     # Block 1
    model.add(layers.Conv2D(8, (3, 3), activation='linear', strides=(2, 2), input_shape=(224, 224, 3)))
    model.add(layers.MaxPooling2D(pool_size = (2,2)))

    # Block 1
    model.add(layers.Conv2D(16, (1, 1), activation='linear', strides=(1, 1)))
    model.add(layers.BatchNormalization())
    model.add(layers.LeakyReLU(alpha=0.3))
    model.add(layers.Conv2D(32, (3, 3), activation='linear', strides=(1, 1)))
    model.add(layers.BatchNormalization())
    model.add(layers.LeakyReLU(alpha=0.3))

    # Block 2
    model.add(layers.Conv2D(32, (3, 3), activation='linear', strides=(1, 1)))
    model.add(layers.LeakyReLU(alpha=0.3))
    model.add(layers.BatchNormalization())

    # Block 3
    model.add(layers.Conv2D(36, (1, 1), activation='linear', strides=(1, 1)))
    model.add(layers.BatchNormalization())
    model.add(layers.LeakyReLU(alpha=0.3))
    model.add(layers.AveragePooling2D(pool_size=(2, 2), strides=(1, 1)))

    # Block 4
    model.add(layers.Conv2D(32, (3, 3), activation='linear', strides=(1, 1)))
    model.add(layers.BatchNormalization())
    model.add(layers.LeakyReLU(alpha=0.3))
    model.add(layers.Conv2D(64, (3, 3), activation='linear', strides=(1, 1)))
    model.add(layers.BatchNormalization())
    model.add(layers.LeakyReLU(alpha=0.3))

    # Block 5
    model.add(layers.Conv2D(64, (3, 3), activation='linear', strides=(1, 1)))
    model.add(layers.BatchNormalization())
    model.add(layers.LeakyReLU(alpha=0.3))


    # Block 6
    model.add(layers.Conv2D(32, (1, 1), activation='linear', strides=(1, 1), kernel_regularizer=regularizers.l2(0.01)))
    model.add(layers.BatchNormalization())
    model.add(layers.LeakyReLU(alpha=0.3))
    model.add(layers.AveragePooling2D(pool_size=(2, 2), strides=(1, 1)))

    # Block 7
    model.add(layers.Conv2D(64, (1, 1), activation='linear', strides=(1, 1), kernel_regularizer=regularizers.l2(0.01)))
    model.add(layers.BatchNormalization())
    model.add(layers.LeakyReLU(alpha=0.3))
    model.add(layers.Conv2D(128, (3, 3), activation='linear', strides=(1, 1), kernel_regularizer=regularizers.l2(0.01)))
    model.add(layers.BatchNormalization())
    model.add(layers.LeakyReLU(alpha=0.3))

    # Block 8
    model.add(layers.Conv2D(128, (3, 3), activation='linear', strides=(1, 1), kernel_regularizer=regularizers.l2(0.01)))
    model.add(layers.BatchNormalization())
    model.add(layers.LeakyReLU(alpha=0.3))

    # Block 9
    model.add(layers.Conv2D(64, (3, 3), activation='linear', strides=(1, 1), kernel_regularizer=regularizers.l2(0.01)))
    model.add(layers.BatchNormalization())
    model.add(layers.LeakyReLU(alpha=0.3))
    model.add(layers.AveragePooling2D(pool_size=(2, 2), strides=(1, 1)))

    # Block 10
    model.add(layers.Conv2D(128, (1, 1), activation='linear', strides=(1, 1), kernel_regularizer=regularizers.l2(0.01)))
    model.add(layers.BatchNormalization())
    model.add(layers.LeakyReLU(alpha=0.3))
    model.add(layers.Conv2D(256, (3, 3), activation='linear', strides=(1, 1), kernel_regularizer=regularizers.l2(0.01)))
    model.add(layers.BatchNormalization())
    model.add(layers.LeakyReLU(alpha=0.3))

    # Block 11
    model.add(layers.Conv2D(256, (3, 3), activation='linear', strides=(1, 1), kernel_regularizer=regularizers.l2(0.01)))
    model.add(layers.BatchNormalization())
    model.add(layers.LeakyReLU(alpha=0.3))


    # Global Average Pooling
    model.add(tf.keras.layers.GlobalAveragePooling2D())
    model.add(layers.Dropout(0.5)) #add dropout 
    
    # Fully Connected Layer
    model.add(Dense(num_classes))

    # Softmax Activation
    model.add(tf.keras.layers.Softmax())

    return model



```

```python
# Display the summary of the LCNet model
lcnet_model = lcnet()
lcnet_model.build((None, 224, 224, 3))  # Manually build the model to display summary
lcnet_model.summary()
```

```python
# Define the filepath for the saved model
checkpoint_filepath = 'best_modelnew1.h5'

# Create a ModelCheckpoint callback
model_checkpoint = ModelCheckpoint(
    filepath=checkpoint_filepath,
    save_best_only=True,  # Only save the best model
    monitor='val_accuracy',  # Monitor validation accuracy
    mode='max',  # Save the model when validation accuracy improves
    verbose=1
)
```

### Compiling the Model

Here I will define parameters to start training with, As the results are evaluated, I might change these parameters to fit achieve a betteer result

```python
# Define hyperparameters and compile the model
learning_rate = 0.001
batch_size = 32
epochs = 100
momentum = 0.99

optimizer = tf.keras.optimizers.Adam(learning_rate=learning_rate)
loss_function = tf.keras.losses.CategoricalCrossentropy()

lcnet_model.compile(optimizer=optimizer, loss=loss_function, metrics=['accuracy'])

# Define early stopping to stop training if validation loss doesn't improve
early_stopping = EarlyStopping(patience=5, monitor='val_accuracy', verbose=1)
```

### Model Training

## Evaluation of Results

Here, I will define a function that will load the best model from the training, evaluate, predict and print the confusion matrix alongside the classification report

```python
def evaluate():
    
    #Plot training and validation loss on y axis and epoch on x axis
    
    plt.plot(history.history['loss'], label='Training Loss')
    plt.plot(history.history['val_loss'], label='Validation Loss')
    #Sets the title of the plot 
    plt.title('Training and Validation Loss')
    #Sets the label of the x-axis 
    plt.xlabel('Epochs')
    #Sets the label of the y-axis 
    plt.ylabel('Loss')
    #Shows the legends of the plot
    plt.legend()
    #Shows the plot on the screen.
    plt.show()
    
    # Plot training and validation accuracy
    plt.plot(history.history['accuracy'], label='Training Accuracy')
    plt.plot(history.history['val_accuracy'], label='Validation Accuracy')
    #Sets the title of the plot 
    plt.title('Training and Validation Accuracy')
    #Sets the label of the x-axis 
    plt.xlabel('Epochs')
    #Sets the label of the y-axis 
    plt.ylabel('Accuracy')
    #Shows the legend of the plot 
    plt.legend()
    #Shows the plot on the screen.
    plt.show()
    
    
    # Evaluate using the best model
    model = load_model(checkpoint_filepath)
    
    # Evaluate the model on the test dataset
    test_loss, test_accuracy = model.evaluate(X_test, y_test, verbose=1)
    
    # Print the test loss and accuracy
    print(f'Test Loss: {test_loss:.4f}')
    print(f'Test Accuracy: {test_accuracy * 100:.2f}%')
    
    #Testing the model
    y_pred_prob =  model.predict(X_test, verbose=1)
    
    # Get the predicted classes (argmax to get the class with the highest probability)
    y_pred = np.argmax(y_pred_prob, axis=1)
    
    # Get the true classes from the test labels
    y_true = np.argmax(y_test, axis=1)
    
    # Generate a classification report
    class_names = os.listdir('/home/750323/Melanoma_Tiles_20k/Tiles')
    print(classification_report(y_true, y_pred, target_names=class_names))
    
    # Print the  Confusion Matrix
    plt.figure(dpi=500, figsize=(50, 20))
    cm = ConfusionMatrixDisplay.from_predictions(y_true,y_pred, display_labels=class_names)
    cm.ax_.tick_params(axis='x', labelsize=10)
    cm.ax_.tick_params(axis='y', labelsize=10)
    cm.ax_.set_xticklabels(cm.ax_.get_xticklabels(), rotation=90)
    plt.title('Confusion Matrix')
    plt.show()
```

```python
#call the evaluate function
evaluate()
```
