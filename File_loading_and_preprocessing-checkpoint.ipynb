## Extraction of Images in the Whole Slide Image Set into Tiles

```python
#Import all necessary libraries
import os #to access directories
import cv2
import json  # to load the annotation which are in coco json format
from shapely.geometry import Polygon, box # needed to cut the images into tiles
import numpy as np  # for preprocessing ans saving as arrays
import pandas as pd #for saving the labels
from rtree import index  # to match the annotations to the image information
import openslide #to read and open the svs images
from PIL import Image # to process images
import random   #to print the samples of preprocessed images
import matplotlib.pyplot as plt #to plot the images
from numpy import *, random
from tqdm import tqdm # to show progress of the processes
```

```python
# Define necessary Constants
BASE_PATH = "/home/750323/Melanoma"  #Directory of the whole slide imaged
ANNOTATIONS_PATH = "/home/750323/CATCH.json" # Annotations
OUTPUT_DIR = "/home/750323/Melanoma_Tiles_20k" # Directory to save the tiles
TILE_SIZE = (224, 224)  # Image size of the tiles, 224 as they will be used in pretrained models that require the size
OVERLAP = 0.2           # To ensure there are no lost information at the edge of the tiles
MAX_TILES = 20000       # Total number of tiles
TILES_PER_IMAGE = MAX_TILES // 50 # Total number of tiles divided by no of images
WHITE_THRESHOLD = 0.5    #threshold for whiteness
```

```python
# Load the annotations
with open(ANNOTATIONS_PATH, 'r') as file:
    annotations_dict = json.load(file)

```

```python

# Check if the directory exists
if os.path.exists(BASE_PATH) and os.path.isdir(BASE_PATH):
    # List all files in the directory
    files = os.listdir(BASE_PATH)
    
    if len(files) > 0:
        print(f"There are {len(files)} files in the directory: {BASE_PATH}")
    else:
        print(f"The directory {BASE_PATH} is empty.")
else:
    print(f"The directory {BASE_PATH} does not exist.")
```

```python

# iteration for each slide in the directory           
for slide_name in os.listdir(BASE_PATH):
    if not slide_name.endswith('.svs'): #the format of the slides
        continue

    slide_path = os.path.join(BASE_PATH, slide_name)

    # Check slide file existence
    if not os.path.exists(slide_path):
        print(f"Warning: Slide {slide_path} not found. Skipping.")
        continue

    # Open slide safely
    try:
        slide = openslide.OpenSlide(slide_path)
    except Exception as e:
        print(f"Error opening {slide_path}. Reason: {e}. Skipping.")
        continue

    # Fetch the image info based on the filename
    image_info = next((img for img in annotations_dict['images'] if img['file_name'] == slide_name), None)
    if not image_info:
        print(f"Image info for slide {slide_name} not found in annotations. Skipping.")
        continue

    # Create an R-tree index for annotations
    idx = index.Index()
    slide_annotations = [a for a in annotations_dict['annotations'] if a['image_id'] == image_info['id']]
    for i, annotation in enumerate(slide_annotations):
        poly = Polygon([(annotation['segmentation'][j], annotation['segmentation'][j+1]) for j in range(0, len(annotation['segmentation']), 2)])
        idx.insert(i, poly.bounds)

    step_size = int(TILE_SIZE[0] * (1 - OVERLAP)) #to define the starting point of the tile extraction and how its going to overlap
    x_steps = (slide.level_dimensions[0][0] - TILE_SIZE[0]) // step_size #steps towards the x coordinate
    y_steps = (slide.level_dimensions[0][1] - TILE_SIZE[1]) // step_size #steps towards the y coordinates
    sampled_tiles = 0

    for x in range(0, x_steps * step_size, step_size): #loop through all the coordinates
        for y in range(0, y_steps * step_size, step_size):
            if sampled_tiles >= TILES_PER_IMAGE:
                break

            tile_bounds = box(x, y, x + TILE_SIZE[0], y + TILE_SIZE[1]) #create like a box to shape the tile
            tile_category = None

            for i in idx.intersection(tile_bounds.bounds): #match the annotations to the images and tiles
                annotation = slide_annotations[i]
                annotation_polygon = Polygon([(annotation['segmentation'][j], annotation['segmentation'][j+1]) for j in range(0, len(annotation['segmentation']), 2)])
                if tile_bounds.intersects(annotation_polygon):
                    tile_category = [cat['name'] for cat in annotations_dict['categories'] if cat['id'] == annotation['category_id']][0]
                    break  #if there are no matching annotation

           # to read the slides and from the lowest region i.e 0 and match the predifined size of 224,224 then converrt to rgb
            if tile_category:
                tile_image = slide.read_region((x, y), 0, TILE_SIZE).convert('RGB')

                # Check the tile's "whiteness"
                tile_np = np.array(tile_image)
                tile_gray = np.mean(tile_np, axis=-1)
                white_fraction = np.sum(tile_gray > 200) / (TILE_SIZE[0] * TILE_SIZE[1])

                if white_fraction < WHITE_THRESHOLD:
                    # Save the tile as a JPEG image
                    tile_output_path = os.path.join(OUTPUT_DIR, "Tiles", tile_category, f"{slide_name}_x{x}_y{y}.jpg")
                    os.makedirs(os.path.dirname(tile_output_path), exist_ok=True)
                    tile_image.save(tile_output_path, "JPEG")


                    sampled_tiles += 1 #for continuous process of the loop
slide.close()
```

#### Processing Tile images as Numpy Array

```python
OUTPUT = "/home/750323/Melanoma_Tiles_20k/Tiles" # The subfolder containing the images

IMG_SIZE = 224
```

```python
mydata = []  #Empty list to append the images and labels
def numpydata():
    # Get the subfolder names dynamically as categories
    categories = [category for category in os.listdir(OUTPUT) if os.path.isdir(os.path.join(OUTPUT, category))]

    # Display progress for all categories
    for category in tqdm(categories, desc="Processing Categories"):
        path = os.path.join(OUTPUT, category)
        class_num = categories.index(category)  # Use index as class number

        # Get the list of images in the category
        images = os.listdir(path)

        for img in images:
            img_array = cv2.imread(os.path.join(path, img))
            new_array = cv2.resize(img_array, (IMG_SIZE, IMG_SIZE))
            mydata.append([new_array, class_num])

numpydata()
```

```python
#To ensure there is relatively even distribution of the data, I will use shuffle
random.shuffle(mydata)
```

```python
X = [] #for the image
Y = [] #for the labels

for features, label in tqdm(mydata, desc="Building NumPy arrays"):
    X.append(features)
    Y.append(label)

X = np.array(X).reshape(-1, IMG_SIZE, IMG_SIZE, 3)
```

```python
#To Normalize the images
X = X.astype('float32')
X /= 255
Print a sample of label and see the shape of the labels
print(Y[100])
print(shape(Y))
```

#### Visualization of the random images in the saved array

```python
# List all the saved image files
num_samples = X.shape[0]
saved_images = list(range(num_samples))

# Randomly select 10 images
random_sample_indices = random.choice(saved_images, size=10, replace=False)

# Define class names
class_names = categories
# Display the selected images with their class names as titles
fig, axes = plt.subplots(2, 5, figsize=(10, 4))
for i, index in enumerate(random_sample_indices):
    row, col = i // 5, i % 5
    image = X[index]
    label_index = Y[index]
    class_name = class_names[label_index]  # Map numeric label to class name
    axes[row, col].imshow(image)
    axes[row, col].set_title(f"{class_name}")
    axes[row, col].axis('off')

plt.tight_layout()
plt.show()
```

```python
# saving the dataset array as .npy file
np.save('X_20k.npy', X)
np.save('Y_20k.npy', Y)
```
